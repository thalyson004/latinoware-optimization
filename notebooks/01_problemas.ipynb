Uma regra if/else seleciona uma ação. Dijkstra resolve exatamente o menor caminho com pesos não negativos. Nem todo problema combinatório é NP-difícil.

```python
import networkx as nx
grafo = nx.Graph()
grafo.add_weighted_edges_from([('Casa','A',2),('A','IFCE',3),('Casa','B',1),('B','IFCE',7)])
print('Regra:', 'sair agora' if 18 < 19 else 'aula iniciou')
print('Menor caminho:', nx.shortest_path(grafo, 'Casa', 'IFCE', weight='weight'))
print('Distância:', nx.shortest_path_length(grafo, 'Casa', 'IFCE', weight='weight'))
```

Com garagem e destino fixos, a ordem de n visitas tem n! possibilidades. A tabela abaixo SUPÕE 1 milhão de rotas completas avaliadas por segundo: não é benchmark de solver. P versus NP permanece aberto.

```python
import math
tabela = []
for n in [5, 10, 15, 20, 25]:
    count = math.factorial(n)
    seconds = count/1_000_000
    tabela.append(dict(visitas=n, rotas=count, segundos=seconds, anos=seconds/(365.25*86400)))
    print(n, f'{count:,} rotas', f'{seconds:.3g} s', f'{seconds/(365.25*86400):.3g} anos')
save('complexity.json', tabela)
```

```python
from pathlib import Path
import sys, json, numpy as np
ROOT = Path.cwd().resolve()
if not (ROOT / 'optimization_examples').is_dir():
    ROOT = ROOT.parent  # notebooks opened from codes/notebooks/
if not (ROOT / 'optimization_examples').is_dir():
    raise RuntimeError('Abra o notebook a partir do repositório codes ou codes/notebooks.')
sys.path.insert(0, str(ROOT))
OUT = ROOT / 'results' / 'sequence'
OUT.mkdir(parents=True, exist_ok=True)
def save(name, value):
    (OUT / name).write_text(json.dumps(value, ensure_ascii=False, indent=2), encoding='utf-8')
```

Dados inteiramente sintéticos. Cada kit é indivisível, disponível uma vez. A prioridade é um escore inventado; não representa benefício social medido. A van admite 15 kg e 12 unidades de volume. Maximize a prioridade respeitando ambos.

```python
from optimization_examples.selection import small_instance, build_model, enumerate_small
instance = small_instance()
for i, nome in enumerate(instance['names']):
    print(nome, 'prioridade', instance['value'][i], 'peso/volume', instance['use'][:,i].tolist())
```

Variáveis: $x_i\in\{0,1\}$. Objetivo: $\max\sum_i v_ix_i$. Regras: $\sum_i a_{ji}x_i\le b_j$ para peso e volume. É programação linear inteira binária (não LP contínua).

```python
model, variables = build_model(instance)
print(model)
reference = enumerate_small(instance)
print('Enumeração completa:', reference)
save('enumeration.json', reference)
```
